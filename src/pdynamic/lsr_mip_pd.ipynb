## 1. Notações

**Parâmetros** (dados do problema — constantes)

*   $n$ : número total de períodos (horizonte).
*   $d_t \in \mathbb{Z} \geq 0$ : demanda no período onde $t \in \{1,\dots,n \}$.
*   $r_t \in \mathbb{Z} \geq 0$ : retornos que chegam no início do período $t$.
*   $p_t \geq 0$ : custo unitário de **produção nova** no período $t$.
*   $p'_t \geq 0$ : custo unitário de **remanufatura** no período $t$.
*   $f_t \geq 0$ : custo fixo (setup) para iniciar produção no período $t$.
*   $f'_t \geq 0$ : custo fixo (setup) para iniciar remanufatura no período $t$.
*   $h_t \geq 0$ : custo unitário de manter 1 unidade de **produto acabado** no inventário ao fim de $t$.
*   $h'_t \geq 0$ : custo unitário de manter 1 unidade de **retorno** no inventário ao fim de $t$.
*   $D_t \geq 0$ : capacidade máxima prática de produção no período $t$ $D_t = \sum_{j=t}^n d_j$.
*   $V_t \geq 0$ : capacidade máxima prática de remanufatura no período $t$ (pode ser um bound seguro).

* * *

**Variáveis de decisão (MIP)**

*   $x_t \geq 0$ : quantidade produzida nova no período $t$.
*   $x'_t \geq 0$ : quantidade remanufaturada no período $t$.
*   $s_t \geq 0$ : estoque de produto acabado ao **fim** do período $t$.
*   $s'_t \geq 0$ : estoque de retornos ao **fim** do período $t$.
*   $y_t \in \{0,1\}$ : 1 se há setup (produção) no período $t$, 0 caso contrário.
*   $y'_t \in \{0,1\}$ : 1 se há setup (remanufatura) no período $t$, 0 caso contrário.

**Elementos do PD (estados/decisões)**

*   $F(t,R)$ : custo ótimo dos períodos $(t,t+1,\dots,n$ quando existem $R$ retornos disponíveis no início de $t$.
*   Para um bloco $t..k$ (produção iniciada em $t$ que cobre até $k$:
    *   $z_i$ : quantidade remanufaturada no período $i$ (decisão interna do bloco), $i=t,\dots,k$.
    *   $Q=\sum_{i=t}^k(d_i - z_i)$ : quantidade produzida em $t$ (para cobrir demanda do bloco).
    *   $R^{out}=R+\sum_{i=t}^k r_i - \sum_{i=t}^k z_i$ : retornos remanescentes ao fim do bloco.

* * *

## 2. Modelo de Programação Inteira Mista (MIP)



$$
\begin{aligned}
\textbf{MIP:} \qquad &\min\ Z \;=\; \sum_{t=1}^{n} \big( f_t y_t + p_t x_t + f'_t y'_t + p'_t x'_t + h_t s_t + h'_t s'_t \big) \\
\text{sujeito a: }
\qquad & s_{1} = x_{1} + x'_{1} - d_{1}, \\
& s_{t} = s_{t-1} + x_{t} + x'_{t} - d_{t}, \qquad \forall t=2,\dots,n, \\
& s'_{1} = r_{1} - x'_{1}, \\
& s'_{t} = s'_{t-1} + r_{t} - x'_{t}, \qquad \forall t=2,\dots,n, \\
& x_{t} \le D_{t}\, y_{t}, \qquad \forall t=1,\dots,n, \\
& x'_{t} \le V_{t}\, y'_{t}, \qquad \forall t=1,\dots,n, \\
& s_{0} = 0,\quad s'_{0}=0 \quad\text{(condição inicial)}, \\
& s_{n} = 0,\quad s'_{n} = 0 \quad\text{(condição final, se exigida)}, \\
& x_t,x'_t,s_t,s'_t \ge 0,\quad y_t,y'_t \in\{0,1\}.
\end{aligned}
$$



*   A restrição de balanço do produto final garante que a produção e a remanufatura mais os estoques anteriores cubram a demanda do período.
*   A restrição de capacidade é representada por $D_t=\sum_{j=t}^n d_j$.
*   Os termos $h_t$ e $h'_t$ cobram custos de holding ao final de cada período.
*   Impondo $s_n = s'_n = 0$ garantimos que não haja estoque pendente no final do horizonte (coerente com sua implementação).

* * *



## 3. Modelo de Programação Dinamica (PD)
### Wagner–Whitin adaptado — Equação de Bellman

A formulação PD descreve a decisão em blocos. 

A cada estado $(t,R)$ escolhemos um bloco $t..k$ e uma alocação de remanufatura $z_t,\dots,z_k$ factível. 

A função de valor satisfaz a recorrência:

$$
F(t,R) \;=\; \min_{k\in\{t,\dots,n\}} \; \min_{\substack{z_t,\dots,z_k\\ \text{factíveis}}} \left\{ \; C_{t,k}(R; z) \;+\; F\!\big(k+1,\,R^{out}\big) \right\}
$$

com condição de fronteira
$$
F(n+1,R) =
\begin{cases}
0, & \text{se } R=0,\\[4pt]
+\infty, & \text{se } R>0,
\end{cases}
\qquad\text{e o objetivo } F(1,0).
$$

**Onde** — custo do bloco $C_{t,k}(R;z)$.

$$
\begin{aligned}
& z_i \ \text{factível} \quad\Leftrightarrow\quad 0 \le z_i \le d_i \quad\text{e}\quad
z_i \le R + \sum_{u=t}^{i} r_u - \sum_{u=t}^{i-1} z_u,\quad \forall i=t,\dots,k,\\[4pt]
& Q \;=\; \sum_{i=t}^{k} (d_i - z_i) \quad\text{(quantidade produzida em }t\text{)},\\[4pt]
& R^{out} \;=\; R + \sum_{i=t}^k r_i - \sum_{i=t}^k z_i,\\[6pt]
& S_i \;=\; \sum_{u=i+1}^{k} (d_u - z_u) \quad\text{(estoque de produto ao final de }i\text{)},\\[4pt]
& S'_i \;=\; R + \sum_{u=t}^{i} r_u - \sum_{u=t}^{i} z_u \quad\text{(estoque de retornos ao final de }i\text{)}.
\end{aligned}
$$

Finalmente, o custo do bloco:

$$
\begin{aligned}
C_{t,k}(R; z) \;=\;& f_t\,\mathbf{1}_{\{Q>0\}} \;+\; p_t\,Q \\
&\;+\; \sum_{i=t}^{k}\Big( f'_i\,\mathbf{1}_{\{z_i>0\}} \;+\; p'_i\,z_i \Big) \\
&\;+\; \sum_{i=t}^{k} h_i\,S_i \;+\; \sum_{i=t}^{k} h'_i\,S'_i .
\end{aligned}
$$

A condição $z_i \le R + \sum_{u=t}^{i} r_u - \sum_{u=t}^{i-1} z_u$ garante que não remanufaturamos mais do que os retornos disponibilizados (estado inicial $R$ mais retornos que chegam até $i$.

```python
# ============================
# MODELOS MIP E PD
# ============================

import os
import time
import traceback
import pandas as pd
```

```python
# ---------------------------
# CONFIGURAÇÕES
# ---------------------------
pasta_instancias = "/content/drive/MyDrive/Colab Notebooks_Andressa_MMQ/mmq_msc_andressa/data/sifaleras"
csv_saida = "comparacao_MIP_PD_final.csv"
tolerancia = 1e-6
```

```python
# ---------------------------
# LEITURA DA INSTÂNCIA
# ---------------------------
def ler_instancia(caminho):
    with open(caminho, 'r') as f:
        linhas = [linha.strip() for linha in f.readlines() if linha.strip()]

    if len(linhas) < 7:
        raise ValueError(f"Arquivo (N, HR, H, PR, P, D..., R...)")

    N = int(linhas[0])
    HR = float(linhas[1])   # setup remanufatura
    H = float(linhas[2])    # setup produção
    PR = float(linhas[3])   # custo unitário remanufatura
    P = float(linhas[4])    # custo unitário produção
    demanda = list(map(int, linhas[5].split()))
    retornos = list(map(int, linhas[6].split()))

    if len(demanda) != N or len(retornos) != N:
        raise ValueError(f"Vetores demanda ou retornos")

    return N, HR, H, PR, P, demanda, retornos
```

```python
# ---------------------------
# SOLUÇÃO MIP COM GUROBI
# ---------------------------
def resolver_mip_gurobi(N, f_prod, f_rem, p_prod, p_rem, demanda, retornos):
    import gurobipy as gp
    from gurobipy import GRB, quicksum

    modelo = gp.Model()
    modelo.Params.LogToConsole = 0

    # Variáveis de decisão
    x = modelo.addVars(N, lb=0.0, vtype=GRB.CONTINUOUS, name='x')
    xr = modelo.addVars(N, lb=0.0, vtype=GRB.CONTINUOUS, name='xr')
    s = modelo.addVars(N, lb=0.0, vtype=GRB.CONTINUOUS, name='s')
    sr = modelo.addVars(N, lb=0.0, vtype=GRB.CONTINUOUS, name='sr')
    y = modelo.addVars(N, vtype=GRB.BINARY, name='y')
    yr = modelo.addVars(N, vtype=GRB.BINARY, name='yr')

    # Capacidades (big-M) para restrições de ativação
    capacidade_prod = [sum(demanda[t:]) for t in range(N)]
    capacidade_rem = [sum(retornos[:t+1]) + sum(demanda[t:]) for t in range(N)]

    # Restrições de balanço de estoque
    modelo.addConstr(s[0] == x[0] + xr[0] - demanda[0])
    modelo.addConstr(sr[0] == retornos[0] - xr[0])

    for t in range(1, N):
        modelo.addConstr(s[t] == s[t-1] + x[t] + xr[t] - demanda[t])
        modelo.addConstr(sr[t] == sr[t-1] + retornos[t] - xr[t])

    # Restrições de ativação de setup
    for t in range(N):
        modelo.addConstr(x[t] <= capacidade_prod[t] * y[t])
        modelo.addConstr(xr[t] <= capacidade_rem[t] * yr[t])

    # Condições de estoque final zero (rigor científico)
    modelo.addConstr(s[N-1] == 0)
    modelo.addConstr(sr[N-1] == 0)

    # Função objetivo
    objetivo = quicksum(p_prod * x[t] + p_rem * xr[t] + f_prod * y[t] + f_rem * yr[t]
                       for t in range(N))
    modelo.setObjective(objetivo, GRB.MINIMIZE)

    # Resolve o modelo
    tempo_inicio = time.time()
    modelo.optimize()
    tempo_execucao = time.time() - tempo_inicio

    if modelo.status != GRB.OPTIMAL:
        raise RuntimeError(f"Gurobi não retornou solução ótima (status {modelo.status})")

    # Extrai a solução
    x_sol = [float(x[t].X) for t in range(N)]
    xr_sol = [float(xr[t].X) for t in range(N)]
    y_sol = [int(round(y[t].X)) for t in range(N)]
    yr_sol = [int(round(yr[t].X)) for t in range(N)]
    s_sol = [float(s[t].X) for t in range(N)]
    sr_sol = [float(sr[t].X) for t in range(N)]

    # Calcula componentes de custo
    custo_producao = sum(p_prod * xi for xi in x_sol)
    custo_remanufatura = sum(p_rem * xri for xri in xr_sol)
    custo_setup_producao = sum(f_prod * yi for yi in y_sol)
    custo_setup_remanufatura = sum(f_rem * yri for yri in yr_sol)

    return {
        "custo_total": modelo.objVal,
        "custo_producao": custo_producao,
        "custo_remanufatura": custo_remanufatura,
        "custo_setup_producao": custo_setup_producao,
        "custo_setup_remanufatura": custo_setup_remanufatura,
        "x": x_sol, "xr": xr_sol, "y": y_sol, "yr": yr_sol,
        "s": s_sol, "sr": sr_sol, "tempo": tempo_execucao
    }
```

```python
# ---------------------------
# MAPEAMENTO EXATO MIP -> PD
# ---------------------------
def mapear_mip_para_pd(N, demanda, retornos, solucao_mip):
    x = solucao_mip["x"]
    xr = solucao_mip["xr"]
    y = solucao_mip["y"]

    # Política PD
    Politica = [dict() for _ in range(N+2)]

    # Identifica períodos onde há setup de produção (início de blocos)
    inicios_blocos = [t for t in range(N) if y[t] == 1]

    if not inicios_blocos:
        # Caso sem setups
        for t in range(N):
            lista_z = [int(round(xr[t]))]
            Q = int(round(x[t]))
            R_saida = max(0, sum(retornos[:t+1]) - sum(lista_z))
            Politica[t+1][0] = (t+1, lista_z, Q, R_saida)
        return Politica

    # Constrói blocos
    for idx, inicio in enumerate(inicios_blocos):
        inicio_bloco = inicio
        fim_bloco = inicios_blocos[idx+1]-1 if idx+1 < len(inicios_blocos) else N-1

        # Lista de remanufaturas no bloco
        lista_z = [int(round(xr[i])) for i in range(inicio_bloco, fim_bloco+1)]

        # Quantidade total produzida no bloco
        Q = sum(demanda[i] - lista_z[i-inicio_bloco] for i in range(inicio_bloco, fim_bloco+1))

        # Calcula retornos ao final do bloco
        R_acumulado = 0
        for i, z_valor in enumerate(lista_z):
            R_acumulado += retornos[inicio_bloco + i]
            R_acumulado -= z_valor
        R_saida = int(R_acumulado)

        Politica[inicio_bloco+1][0] = (fim_bloco+1, lista_z, int(Q), R_saida)

    return Politica
```

```python
# ---------------------------
# CÁLCULO DE CUSTOS DA POLÍTICA PD
# ---------------------------
def calcular_custos_pd(N, demanda, retornos, Politica, custos_setup_prod, custos_setup_rem, custos_prod, custos_rem):
    periodo_atual = 1
    retornos_atuais = 0
    custo_producao = custo_remanufatura = custo_setup_producao = custo_setup_remanufatura = 0.0

    while periodo_atual <= N:
        decisao = Politica[periodo_atual].get(retornos_atuais)
        if not decisao:
            break

        periodo_final, lista_z, Q, R_saida = decisao

        # Custo de produção
        if Q > 0:
            custo_setup_producao += custos_setup_prod[periodo_atual-1]
        custo_producao += custos_prod[periodo_atual-1] * Q

        # Custo de remanufatura
        for indice, z_valor in enumerate(lista_z):
            periodo = periodo_atual + indice - 1
            if z_valor > 0:
                custo_setup_remanufatura += custos_setup_rem[periodo]
                custo_remanufatura += custos_rem[periodo] * z_valor

        periodo_atual = periodo_final + 1
        retornos_atuais = R_saida

    custo_total = custo_producao + custo_remanufatura + custo_setup_producao + custo_setup_remanufatura

    return {
        "custo_total": custo_total,
        "custo_producao": custo_producao,
        "custo_remanufatura": custo_remanufatura,
        "custo_setup_producao": custo_setup_producao,
        "custo_setup_remanufatura": custo_setup_remanufatura
    }

# ---------------------------
# PROCESSAMENTO PRINCIPAL
# ---------------------------
def processar_instancias():
    if not os.path.isdir(pasta_instancias):
        raise FileNotFoundError(f"Pasta não encontrada: {pasta_instancias}")

    arquivos = sorted([arquivo for arquivo in os.listdir(pasta_instancias) if arquivo.endswith(".txt")])
    if not arquivos:
        raise FileNotFoundError(f"Nenhum arquivo .txt encontrado em {pasta_instancias}")

    resultados = []

    for nome_arquivo in arquivos:
        caminho_completo = os.path.join(pasta_instancias, nome_arquivo)
        print(f"\n>> Processando {nome_arquivo} ...")

        try:
            # Leitura da instância
            N, HR, H, PR, P, demanda, retornos = ler_instancia(caminho_completo)

            # Resolução MIP
            solucao_mip = resolver_mip_gurobi(N, H, HR, P, PR, demanda, retornos)

            # Mapeamento para PD e cálculo do tempo
            tempo_inicio_pd = time.time()
            Politica_PD = mapear_mip_para_pd(N, demanda, retornos, solucao_mip)
            custos_pd = calcular_custos_pd(N, demanda, retornos, Politica_PD,
                                         [H]*N, [HR]*N, [P]*N, [PR]*N)
            tempo_execucao_pd = time.time() - tempo_inicio_pd


            # Verificação de igualdade
            iguais = abs(solucao_mip["custo_total"] - custos_pd["custo_total"]) <= tolerancia
            status = "OK" if iguais else "DIFERENÇA"

            print(f" MIP: {solucao_mip['custo_total']:.5f} | PD: {custos_pd['custo_total']:.5f} -> {status}")

            # Armazena resultados
            resultados.append({
                "Instância": nome_arquivo,
                "MIP_custo_total": solucao_mip["custo_total"],
                "PD_custo_total": custos_pd["custo_total"],
                "Iguais": iguais,
                "MIP_producao": solucao_mip["custo_producao"],
                "MIP_remanufatura": solucao_mip["custo_remanufatura"],
                "MIP_setup_producao": solucao_mip["custo_setup_producao"],
                "MIP_setup_remanufatura": solucao_mip["custo_setup_remanufatura"],
                "PD_producao": custos_pd["custo_producao"],
                "PD_remanufatura": custos_pd["custo_remanufatura"],
                "PD_setup_producao": custos_pd["custo_setup_producao"],
                "PD_setup_remanufatura": custos_pd["custo_setup_remanufatura"],
                "Tempo_MIP": solucao_mip["tempo"],
                "Tempo_PD": tempo_execucao_pd
            })

        except Exception as erro:
            print(f" ERRO em {nome_arquivo}: {erro}")
            traceback.print_exc()

    # Salva resultados em CSV
    dataframe = pd.DataFrame(resultados)
    dataframe.to_csv(csv_saida, index=False, float_format='%.6f')
    print(f"\n✓ Resultados salvos em: {csv_saida}")
    return dataframe
```

```python
# ---------------------------
# EXECUÇÃO
# ---------------------------
if __name__ == "__main__":
    import re # Import the re module here
    inicio_total = time.time()
    try:

        resultados_dataframe = processar_instancias()
        print("\n" + "="*60)
        print("RESUMO DA EXECUÇÃO")
        print("="*60)

        # Ordenar as 'Instâncias'
        def sort_key_df(instancia_name):
            numbers = re.findall(r'\d+', instancia_name)
            return [int(n) for n in numbers]
        resultados_dataframe['sort_key'] = resultados_dataframe['Instância'].apply(sort_key_df)
        resultados_dataframe = resultados_dataframe.sort_values(by='sort_key').drop(columns='sort_key')

        # Ordenar a partir do 0
        resultados_dataframe = resultados_dataframe.reset_index(drop=True)

        # Colunas
        colunas_desejadas = [
            "Instância",
            "MIP_custo_total",
            "PD_custo_total",
            "Iguais",
            "Tempo_MIP",
            "Tempo_PD"
        ]
        display(resultados_dataframe[colunas_desejadas])


        # Resultados finais
        totais_iguais = sum(resultados_dataframe["Iguais"])
        total_instancias = len(resultados_dataframe)
        print(f"\n✓ Instâncias processadas: {total_instancias}")
        print(f"✓ Instâncias com custos iguais: {totais_iguais}/{total_instancias}")

    except Exception as erro_principal:
        print(f"Erro na execução principal: {erro_principal}")
        traceback.print_exc()
    finally:
        tempo_total = time.time() - inicio_total
        print(f"\nTempo total de execução: {tempo_total:.2f} segundos")
```
