# Formulação por Programação Dinâmica do PDLR-R

A Programação Dinâmica modela o PDLR-R como um processo de decisão sequencial determinístico, no qual decisões produtivas são tomadas ao longo de um horizonte finito de planejamento.

## Estágios de decisão

O horizonte de planejamento é dividido em períodos discretos:

$
t = 1, 2, \dots, n,
$

em que cada período corresponde a um estágio de decisão do problema.

## Estado

O estado do sistema no início do período \(t\) é definido pelo vetor:

($s_{t-1}^p, s_{t-1}^r$)

onde:

- $s_{t-1}^p$ representa o estoque de produtos manufaturados disponíveis no início do período \(t\);
- $s_{t-1}^r$ representa o estoque de itens retornados disponíveis para remanufatura no início do período \(t\).

## Função de transição de estados

A evolução do sistema é descrita pelas equações de balanço de estoque:

\begin{align}
s_t^p &= s_{t-1}^p + x_t^p + x_t^r - d_t, \\
s_t^r &= s_{t-1}^r + r_t - x_t^r,
\end{align}

## Função de custo imediato
$$
g_t(s_{t-1}^p,s_{t-1}^r,x_t^p,x_t^r) = h_t^p s_t^p + h_t^r s_t^r + p_t^p x_t^p + p_t^r x_t^r + f_t^p \cdot \mathbf{1}(x_t^p > 0) + f_t^r \cdot \mathbf{1}(x_t^r > 0)
$$
onde $\mathbf{1}(\cdot)$ é a função indicadora:

\[
\mathbf{1}(\text{condição}) = \begin{cases} 1, & \text{se a condição é satisfeita}, \\ 0, & \text{caso contrário}. \end{cases}
\]

```python
import pandas as pd
import numpy as np
import os
import time
from typing import Dict, Tuple, List
```

```python
#Estados
State = Tuple[int, int]           # (s_novo, s_reman)
Decision = Tuple[int, int]        # (x_novo, x_reman)
```

```python
#Parâmetros
class HeuristicParams:
    """
    Parâmetros para modular heurísticas e poda.
    Ajuste fina aqui para trocar qualidade x velocidade.
    """
    def __init__(
        self,
        lookahead_horizons: List[int] = [2, 3, 5, 7],
        reman_sampling_small_range_div: int = 10,
        reman_sampling_large_range_div: int = 20,
        reman_sampling_threshold: int = 50,
        max_states: int = 8000
    ):
        self.lookahead_horizons = lookahead_horizons
        self.reman_sampling_small_range_div = reman_sampling_small_range_div
        self.reman_sampling_large_range_div = reman_sampling_large_range_div
        self.reman_sampling_threshold = reman_sampling_threshold
        self.max_states = max_states
```

```python
# Ler arquivo
def read_datafile(path: str):
    """
    Lê arquivo no formato esperado (mesmo do seu original).
    Retorna: T, FR, F, HR, H, D (np.array), R (np.array)
    """
    with open(path, 'r') as f:
        linhas = [ln.strip() for ln in f.readlines()]

    idx = 0
    T = int(linhas[idx]); idx += 1

    FR = float(linhas[idx]); idx += 1
    F = float(linhas[idx]); idx += 1
    HR = float(linhas[idx]); idx += 1
    H = float(linhas[idx]); idx += 1

    D = np.fromstring(linhas[idx], dtype=np.int64, sep='\t'); idx += 1
    R = np.fromstring(linhas[idx], dtype=np.int64, sep='\t')

    return T, FR, F, HR, H, D.astype(int), R.astype(int)
```

```python
# Geração de decisões
# ---------------------------
def generate_decisions(
    s: int,
    sr: int,
    d: int,
    r: int,
    D: np.ndarray,
    t: int,
    T: int,
    params: HeuristicParams
) -> List[Decision]:
    """
    Retorna uma lista de decisões (x_novo, x_reman) razoáveis para o estado (s, sr) no período t.

     - não produzir se já há estoque
     - produzir exatamente o necessário (novo ou reman)
     - mix "usar reman primeiro"
     - lookahead Wagner-Whitin (horizontes em params)
     - amostragem de remanufatura entre need e available_r (grade adaptativa)
    """

    decisions: List[Decision] = []

    # demanda do período
    need = max(0, d - s)

    # limites teóricos (máxima produção necessária para esvaziar demanda restante até o fim)
    # Notar: D_cum pode ser calculado no chamador, mas aqui usamos lookahead local.
    # available remanufatura
    available_r = sr + r

    is_last = (t == T - 1)

    # 1) Não produzir (válido se s >= d)
    if s >= d:
        decisions.append((0, 0))

    # 2) Produzir apenas o necessário (novos ou reman)
    if need > 0:
        # produzir só novos
        decisions.append((need, 0))
        # produzir só reman (se houver disponibilidade)
        if need <= available_r:
            decisions.append((0, need))

        # 3) Mix: preferir reman quando possível (usar reman primeiro)
        reman_use = min(need, available_r)
        novo_use = need - reman_use
        decisions.append((novo_use, reman_use))

    # 4) Wagner-Whitin lookahead: produzir para atender demanda futura acumulada
    # (apenas quando não é último período)
    if not is_last:
        for h in params.lookahead_horizons:
            if t + h > T:
                continue
            fut_demand = int(np.sum(D[t : min(t + h, T)]))
            total_needed = max(0, fut_demand - s)
            if total_needed > 0:
                decisions.append((total_needed, 0))
                if total_needed <= available_r:
                    decisions.append((0, total_needed))

    # 5) Exploração da remanufatura: amostragem adaptativa entre need e available_r
    if need > 0 and available_r > 0 and not is_last:
        low = need
        high = available_r
        # pontos críticos: necessidade, meio, disponibilidade
        mid = (low + high) // 2
        test_points = {low, mid, high}

        # grade adaptativa
        span = high - low
        if span <= params.reman_sampling_threshold:
            step = max(1, span // params.reman_sampling_small_range_div)
        else:
            step = max(1, span // params.reman_sampling_large_range_div)

        q = low
        while q <= high:
            test_points.add(q)
            q += step

        # adicionar decisões (x_novo = 0, x_reman = q) válidas
        for q in sorted(test_points):
            if q >= 0:
                decisions.append((0, q))

    # Evitar duplicatas exatas mantendo ordem (conservadora)
    seen = set()
    unique_decisions = []
    for dec in decisions:
        if dec not in seen:
            seen.add(dec)
            unique_decisions.append(dec)

    return unique_decisions
```

```python
# ---------------------------
# Avaliar transição e custo
# ---------------------------
def evaluate_transition(
    s: int,
    sr: int,
    x_novo: int,
    x_reman: int,
    d: int,
    r: int,
    base_cost: float,
    F: float,
    FR: float,
    H: float,
    HR: float,
    is_last: bool
):
    """
    Avalia a transição e retorna (new_s, new_sr, new_cost)
    Se decisão inviável, retorna (None, None, +inf)
    """

    # Validações simples
    if x_novo < 0 or x_reman < 0:
        return None, None, np.inf

    new_s = s + x_novo + x_reman - d
    new_sr = sr + r - x_reman

    # estoques não negativos
    if new_s < 0 or new_sr < 0:
        return None, None, np.inf

    # último período - forçar new_s == 0 (opcional: pode ser relaxado)
    if is_last and new_s != 0:
        return None, None, np.inf

    new_cost = base_cost
    # setups
    if x_novo > 0:
        new_cost += F
    if x_reman > 0:
        new_cost += FR

    # holding
    if is_last:
        # No seu código original, havia tratamento especial:
        new_cost += H * new_s
        # "não penalizar retornos inevitáveis do último período" -> aproximamos:
        leftover = max(0, new_sr - r)  # conforme original
        new_cost += HR * leftover
    else:
        new_cost += H * new_s + HR * new_sr

    return new_s, new_sr, new_cost
```

```python
# Poda de estados (mantém os melhores por custo)
# ---------------------------
def prune_states(states: Dict[State, float], max_states: int) -> Dict[State, float]:
    """
    Mantém apenas os melhores 'max_states' estados com menor custo.
    Retorna novo dicionário podado.
    """
    if len(states) <= max_states:
        return states

    # Ordena por custo e mantém os primeiros max_states
    items = sorted(states.items(), key=lambda kv: kv[1])
    pruned = dict(items[:max_states])
    return pruned
```

```python
# ---------------------------
# DP principal com backtracking
# ---------------------------
def ulsr_dp_readable(
    datafile: str,
    params: HeuristicParams = None,
    verbose: bool = True
):
    """
    Executa o DP legível com reconstrução da política ótima.
    Retorna um dict com:
      - best_cost
      - best_final_state
      - policy: lista de decisões por período (t=0..T-1)
      - diagnostics (tempos, num estados por período)
    """

    if params is None:
        params = HeuristicParams()

    # Ler dados
    T, FR, F, HR, H, D, R = read_datafile(datafile)

    if verbose:
        print("ULSR DP (legível)  iniciando")
        print(f"T={T}, F={F}, FR={FR}, H={H}, HR={HR}")
        print(f"£D={int(D.sum())}, £R={int(R.sum())}")

    # Demanda acumulada (útil para limites)
    D_cum = np.zeros(T + 1, dtype=int)
    for i in range(T - 1, -1, -1):
        D_cum[i] = D[i] + D_cum[i + 1]

    # Estado inicial
    states: Dict[State, float] = {(0, 0): 0.0}   # custo até aqui
    # Predecessores para backtracking:
    # list com um dict por período: pred[t][state] = (prev_state, decision)
    preds: List[Dict[State, Tuple[State, Decision]]] = []

    diagnostics = {
        'num_states_per_period': [],
        'time_per_period': []
    }

    start_total = time.time()

    # Forward DP
    for t in range(T):
        t_start = time.time()
        d = int(D[t]); r = int(R[t])
        is_last = (t == T - 1)

        new_states: Dict[State, float] = {}
        pred_this_period: Dict[State, Tuple[State, Decision]] = {}

        # Para cada estado atual, gerar decisões e atualizar
        for state, base_cost in states.items():
            s, sr = state

            # limites práticos (pode usar D_cum para maior restrição)
            max_novo = max(0, D_cum[t] - s)       # não precisamos produzir mais que demanda restante
            max_reman = max(0, min(sr + r, D_cum[t] - s))

            # gerar decisões relevantes
            decisions = generate_decisions(s, sr, d, r, D, t, T, params)

            for dec in decisions:
                x_novo, x_reman = dec

                # ignorar rapidamente decisões que excedem limites simples
                if x_novo > max_novo or x_reman > (sr + r):
                    continue

                new_s, new_sr, new_cost = evaluate_transition(
                    s, sr, x_novo, x_reman,
                    d, r,
                    base_cost,
                    F, FR, H, HR,
                    is_last
                )

                if new_s is None:
                    continue

                key = (int(new_s), int(new_sr))

                # atualizar custo ótimo para esse novo estado
                if key not in new_states or new_cost < new_states[key]:
                    new_states[key] = new_cost
                    pred_this_period[key] = (state, dec)

        # poda de estados
        if len(new_states) > params.max_states:
            new_states = prune_states(new_states, params.max_states)
            # remover entradas de pred_this_period relativas a estados podados
            pred_this_period = {k: v for k, v in pred_this_period.items() if k in new_states}

        states = new_states
        preds.append(pred_this_period)

        t_end = time.time()
        diagnostics['num_states_per_period'].append(len(states))
        diagnostics['time_per_period'].append(t_end - t_start)

        if verbose and (t % 5 == 0 or t == T - 1):
            print(f" t={t+1}/{T} | estados={len(states):,} | tempo={t_end-t_start:.3f}s")

        # Caso não haja estados viáveis, saímos
        if not states:
            if verbose:
                print("Nenhum estado viável encontrado em t=", t)
            break

    total_time = time.time() - start_total

    # ---------------------------
    # Encontrar estado final (s == 0) com menor custo
    # ---------------------------
    best_cost = np.inf
    best_state = None
    for st, c in states.items():
        s, sr = st
        if s == 0 and c < best_cost:
            best_cost = c
            best_state = st

    if best_state is None:
        # fallback: pegar menor custo qualquer (se condição s==0 for muito restritiva)
        for st, c in states.items():
            if c < best_cost:
                best_cost = c
                best_state = st
    # Backtracking para reconstruir política ótima
    # ---------------------------
    policy: List[Decision] = [ (0,0) ] * T   # inicializa com decisões neutras
    cur_state = best_state

    # Recorremos os preds de trás para frente
    for t in range(T - 1, -1, -1):
        if t >= len(preds):
            # não tínhamos preds para este período (falha)
            break
        pred_dict = preds[t]
        if cur_state not in pred_dict:
            break
        prev_state, decision = pred_dict[cur_state]
        policy[t] = decision
        cur_state = prev_state

    # montar resultado
    result = {
        'best_cost': float(best_cost),
        'best_final_state': best_state,
        'policy': policy,
        'diagnostics': {
            **diagnostics,
            'total_time': total_time,
            'T': T
        },
        'objbound': float(best_cost), # DP best_cost
        'mipgap': 0.0,
        'nodes': np.nan
    }

    if verbose:
        print("\nResultado:")
        print(f" best_cost = {best_cost:.4f}")
        print(f" best_final_state = {best_state}")
        print(f" total_time = {total_time:.3f}s")
        print(f" estados por período (sample): {diagnostics['num_states_per_period'][:10]} ...")

    return result
```

```python
results = []
base_data_dir = "/content/drive/MyDrive/MMQ_MSC_ANDRESSA/mmq_msc_andressa/data/sifaleras/"
params = HeuristicParams()

for dim in [52]:
    for id in range(1, 11):
        datafile_name = f"{dim}_{id}.txt"
        datafile_full_path = f"{base_data_dir}{datafile_name}"

        # Executa o DP e armazena os resultados
        out = ulsr_dp_readable(datafile_full_path, params=params, verbose=False)

        results.append({
            'instance': datafile_name,
            'objval': out['best_cost'],
            'time': out['diagnostics']['total_time'],
            'opt': 1  #sucesso da DP = 1
        })

tab = pd.DataFrame(results)

# Resultados do DP
resume = pd.DataFrame({
    'instance': ['resume'],
    'objval': [tab['objval'].mean()],
    'time': [tab['time'].mean()],
    'opt': [tab['opt'].sum()]
})

tab = pd.concat([tab, resume], ignore_index=True)

tab['objval'] = tab['objval'].round(2)
tab['time'] = tab['time'].round(2)
tab['opt'] = tab['opt'].astype('Int64')
```
